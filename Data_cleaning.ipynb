```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/cc46c4cc-36dd-4726-8f7a-410a99f91d69_Data.csv")

df = df.drop(columns=["Series Name", "Series Code"], errors="ignore")
df = df.melt(id_vars=["Country Name", "Country Code"], var_name="Year", value_name="Methane")
df = df.dropna()
df["Year"] = df["Year"].str.extract(r'(\d{4})').astype(int)
df["Methane"] = pd.to_numeric(df["Methane"], errors="coerce")
df = df.dropna(subset=["Methane"])

exclude_codes = [
    "WLD","IBD","IBT","IDA","IDX","MIC","HIC","LIC","LMC","UMC",
    "EAS","ECS","LCN","MEA","NAC","SAS","SSF","EUU","EMU",
    "ARB","CEB","CSS","EAP","EAR","ECA","FCS","LAC","MNA","OED",
    "PSS","PST","PRE","SST","TEA","TEC","TLA","TMN","TSA","TSS"
]

exclude_names = [
    "World","High income","Low income","Middle income","Low & middle income",
    "Lower middle income","Upper middle income",
    "East Asia & Pacific","Europe & Central Asia","Latin America & Caribbean",
    "Middle East & North Africa","North America","South Asia","Sub-Saharan Africa",
    "European Union","Euro area","OECD members","IDA total","IDA blend","IDA only",
    "IBRD only","IDA & IBRD total","Late-demographic dividend",
    "Early-demographic dividend","Post-demographic dividend",
    "Pre-demographic dividend","Pacific island small states",
    "Small states","Other small states","Caribbean small states"
]

df_clean = df[
    (~df["Country Code"].isin(exclude_codes)) &
    (~df["Country Name"].isin(exclude_names))
]

print(df_clean[["Country Name","Country Code"]].drop_duplicates().head(20))
print(df_clean["Country Name"].nunique())
```

```python
# Remove non-country names using keywords
exclude_keywords = [
    "income", "dividend", "Asia", "Europe", "Africa", "America",
    "Caribbean", "Pacific", "OECD", "IDA", "IBRD", "World",
    "Euro", "Union", "states"
]

df_clean = df_clean[
    ~df_clean["Country Name"].str.contains('|'.join(exclude_keywords), case=False, na=False)
]

print(df_clean["Country Name"].unique()[:20])
```

```python
df_clean = df[
    (df["Country Code"].str.len() == 3) &          # ISO country codes
    (~df["Country Name"].str.contains(
        "income|dividend|countries|situations|classified|OECD|IDA|IBRD|World|Asia|Africa|Europe|America|Caribbean|Pacific",
        case=False, na=False
    ))
]
```
